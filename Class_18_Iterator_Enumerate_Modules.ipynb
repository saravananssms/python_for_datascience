# Enumerator

```python
a = 'python'

list(enumerate(a))
```

```python
a = 'this is a pythons class'

for i,j in enumerate(a):
    if j == 's':
        print(i)
```

```python
a = ['a' , 34 , 'a' , 56 , 'a']

tuple(enumerate(a ))
```

# Special char ---> escape char

```python
# \n ---> new line


a = 'hello\nworld'

print(a)
```

```python
# \t ---> Tab space


a = 'hello\tworld'

print(a)
```

```python
# \b ---> Backspace
# Deletes the previous char


a = 'hello\bworld'

print(a)
```

```python
hellworld
```

```python
# Escape char --> back slash


a = 'hello\nworld\tpython'

print(a)
```

```python
a = 'hello\\nworld\tpython'

print(a)
```

# Raw string

```python
a = r'hello\nworld\tpython'

print(a)
```

# strip , rstrip , lstrip

- removes the specified char from the end of the strings
- If specified char is not mentioned by default it'll remove the space

```python
a = '              python                                    '


print(a)
```

```python
a = '              python                                    '


print(a.strip())
```

```python
a = '*************python***********************************'


print(a.strip('*'))
```

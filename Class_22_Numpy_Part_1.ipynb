# Iterating over an array

```python
a = np.arange(36).reshape(6,-1)

a
```

```python
for i in a:
    print(i , end = '\n\n\n')
```

```python
for i in a:
    for j in i:
        print(j)
```

```python
# Ravel , flatten
```

```python
a.ravel()
```

```python
a.flatten()
```

```python
for i in a.ravel():
    print(i)
```

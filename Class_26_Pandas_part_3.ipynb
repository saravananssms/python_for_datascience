# Concatination

```python
india_weather = pd.DataFrame({
    "city": ["mumbai","delhi","banglore"],
    "temperature": [32,45,30],
    "humidity": [80, 60, 78]
})
india_weather
```

```python
us_weather = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68, 65, 75]
})
us_weather
```

```python
pd.concat([india_weather , us_weather] , ignore_index= True)
```

```python
pd.concat([india_weather , us_weather]  , keys= ['INDIA' , 'US'])
```

```python
del us_weather['humidity']
```

# Merge

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","orlando"],
    "humidity": [65,68,75],
})
df2
```

```python
pd.concat([df1 , df2])
```

```python
pd.merge(df1 , df2 , on = 'city' ) # how='inner' , on=None
```

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35, 38],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "humidity": [65,68,71],
})
df2
```

```python
pd.merge(df1 , df2 , on = 'city' )
```

```python
pd.merge(df1 , df2 , on = 'city' , how = 'left')
```

```python
pd.merge(df1 , df2 , on = 'city' , how = 'right')
```

```python
pd.merge(df1 , df2 , on = 'city' , how = 'outer')
```

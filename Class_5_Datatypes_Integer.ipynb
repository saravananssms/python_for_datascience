# Comparision operators

< , > , >= , <=

- The output will always be an bool
- Can be performed on int - float
- Cannot perform on complex data type
- Cannot perform on None data type
- Cannot perform on numbers to string
- Can perform comparision operators on str to str

```python
a = 10
b = 45


print(a < b)
```

```python
a = 10
b = 45


print(a > b)
```

```python
a = 10
b = 10


print(a >= b)
```

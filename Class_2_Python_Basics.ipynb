# Data types in python


1. Fundamental data types
- int - Any number without a decimal point eg -100 , 0 , 100
- Float - Any number with a decimal point -100.023 , 0.0 , 3434.0
- complex data types - Anything that can be written in a form of a+bj where a is real and b is the imag
- String data type - Anything written inside '..' , ".." , '''..''' , """.."""
- Bool --> True / False
- None

2. Derived data types
- list
- tuple
- dic
- set

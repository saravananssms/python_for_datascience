# Aliasing

```python
a = [1,2,3,4]

b = a # Aliasing


print(a)
print(b)


print('--------------')
print(id(a))
print(id(b))


b[0] = 999999


print('---------')

print(a)
print(b)
```

# Shallow copy

```python
print(dir(list))
```

```python
a = [1,2,3,4]

b = a.copy() # Shallow copy


print(a)
print(b)


print('--------------')
print(id(a))
print(id(b))


b[0] = 999999


print('---------')

print(a)
print(b)
```

# Failing point of shallow copy

```python
a = [1,2,[3,4]]

b = a.copy() # Shallow copy


print(a)
print(b)


print('--------------')
print(id(a))
print(id(b))


b[0] = 999999


print('---------')

print(a)
print(b)
```

```python
a = [1,2,[3,4]]

b = a.copy() # Shallow copy


print(a)
print(b)


print('--------------')
print(id(a[-1]))
print(id(b[-1]))


b[-1][0] = 999999


print('---------')

print(a)
print(b)
```

# Deep copy

```python
from copy import deepcopy
```

```python
a = [1,2,[3,4]]

b = deepcopy(a)


print(a)
print(b)


print('--------------')
print(id(a))
print(id(b))

print('--------------')
print(id(a[-1]))
print(id(b[-1]))


b[-1][0] = 999999


print('---------')

print(a)
print(b)
```

```python
# Indexing - Extracting one char at a time


### 2 Types of index values

- +ve ---> L- R --> starts from 0
- -ve ---> R-L ---> Starts from -1


syntax indexing

var[index value]
```

```python
a = 'python'


print(a[2])

print(a[-4])
```

# Slicing
- Extracting multiple char at a time

- Start --> Inclusive
- Stop --> Exclusive
syntax

var[start : stop : step]


If the step is +ve --> L- R
If the step is negative ----> R - L

### Default values

a[:] & a[::1]
start = 0
stop = len(str)



a[::-1]
start = -1
stop = - (len(str)+1)

```python
a = 'python'

a[2:4]
```

```python
date = '24-07-2023'

#DATE
print(date[0:2])
print(date[-10:-8])
print(date[0:-8])
print(date[-10:2])

#MONTH
print(date[3:5])
print(date[-7:-5])
print(date[3:-5])
print(date[-7:5])

#YEAR
print(date[6:])
print(date[6:len(date)])
print(date[6:10])
print(date[-4:])
print(date[-4:len(date)])
print(date[-4:10])
```

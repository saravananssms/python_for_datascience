# Showing errors to the user

```python
a = 10
b = 0

try:
    print(a / b)
    
except Exception as e:
    print(e)
    print('back up code')

print('Next function')
```

```python
a = 10
b = 'abc'

try:
    print(a / g)
    
except Exception as e:
    print(e)
    print('back up code')

print('Next function')
```

```python
a = 10
b = 'abc'

try:
    print(a / g)
    
except ZeroDivisionError as e:
    print(e)
    print('back up code for Zero division')
    
except TypeError as e:
    print(e)
    print('back up code for Type error')
    
# except Exception as e:
#     print(e)
#     print('common back up code')

print('Next function')
```

```python
a = 10
b = 0

try:
    print(a / b)
    
except ZeroDivisionError:
    print('back up code for Zero division')
    
except TypeError as e:
    print(e)
    print('back up code for Type error')
    
except Exception as e:
    print(e)
    print('common back up code')

print('Next function')
```

# Properties of dict 

- It is a collection of key and a value pair
- A key and a values are seperated by :
- A key and a value together is known as an item
- Each Item is seperated by ,
- A key cannot be duplicated
- Whereas a value can be duplicated
- A key cannot be an mutable data type
- Whereas a value can be mutable / immutable
- It is not a seq data type hence no indexing, slicing , concationation , repetation
- Membership operators are possible but only on the keys
- Identity operators are possible

# Variable length arguments

*args , **kwargs

- *args --> It takes the variable length positional arguments and store it in a form oa a tuple
- **kwargs --> Takes the variable length keyword arguments and store it in a form of dict

```python
def adder(a,b):
    print(a+b)
```

```python
adder(10,34,34,45,56,67,67,78,67,564)
```

```python
def adder(*args):
    print(args , type(args))
    print(sum(args))
```

```python
adder(10,20,30)
```

```python
def adder(**kwargs):
    print(kwargs , type(kwargs))
```

```python
adder(hello = 45 , world = 56 , a = 56)
```

```python
dict(a = 345 , b = 45)
```

```python
def adder(*args , **kwargs):
    print(args , type(args))
    print(kwargs , type(kwargs))
```

```python
adder(23,34,345,45,hello = 45 , world = 56 , a = 56)
```

```python
emp_data = {'emp_1' : {'Name':'Salahuddin' , 'Age':32 , 'Salary':46554656} , 
    'emp_2' : {'Name':'Pratik' , 'Age':35 , 'Salary':545454} , 
    'emp_3' : {'Name':'Dilip' , 'Age':45 , 'Salary':65656565} , 
    'emp_4' : {'Name':'Ajaya' , 'Age':56 , 'Salary':6767676} , 
    'emp_5' : {'Name':'Akash' , 'Age':23 , 'Salary':45454545}}
```

```python
def emp_adder(name , age , salary):
    key = f"emp_{len(emp_data)+1}"
    emp_data.setdefault(key , {'Name':name , 'Age':age , 'Salary':salary})
```

```python
emp_adder('xyz' , 34 , 5656565 , pin = 3434343)
```

```python
emp_data
```

```python
def emp_adder(name , age , salary,location = 'Mumbai' , **kwargs):
    key = f"emp_{len(emp_data)+1}"
    details = {'Name':name , 'Age':age , 'Salary':salary , 'location': location}
    details.update(kwargs)
    emp_data.setdefault(key , details)
```

```python
emp_adder('new' , 34 , 5656565 , pan = 4545454)
```

# Vector stacking , concatenation

```python
a = np.arange(12).reshape(6,-1)**2
a
```

```python
b = np.arange(12).reshape(6,-1)**5
b
```

```python
np.concatenate((a,b)) #  axis=0 --> row wise concat
```

```python
np.vstack((a,b))
```

```python
np.concatenate((a,b) , axis = 1) #  axis=1-->columns
```

```python
np.hstack((a,b))
```

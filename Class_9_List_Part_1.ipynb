# Properties of list


- It is a seq data type
- It allows both +ve and -ve indexing
- Perform slicing
- It allows duplication
- It is hetrogeneous ---> Can store different data types
- basic op concatination , repetation , Identity , Membership


##### A list is an mutable 

```python
# Indexing 


a = [34,45,56,56,56,56,56,6,78,89]


print(a[2])


print(a[-4])
```

```python
# Slicing



a = [34,45,56,6,78,89]
```

```python
# 45,56,6

a[1:4]
```

```python
# 6,56,45

a[-3:-6:-1]
```

```python
# It allows duplication

lst = ['a' , 'b' , 'c' , 'a' , 'a' , 'a']

lst
```

```python
# It is hetrogeneous ---> Can store different data types



lst = ['a' ,345,65.56,34+45j , [4,5,67,7] , {'a':456}]

lst
```

```python
# Concatination
# op --> +
# operand --> Both Should be a list


a = [1,2,3]
b = [4,5,6]

print(a+b)
```

```python
# Repetation
# op --> *
# operand --> One should be an int and other should be a list


a = [1,2,3]
b = 2 

print(a * b)
```

```python
lst = [1,2,3,4,5]

lst * 5
```

```python
re = [5,10,13,20,25]
```

```python
# Membership op
# in and not in



lst = [1,2,3,4,5]


3 in lst
```

```python
lst = [1,2,3,4,5]


3 not in lst
```

```python
lst = [1,2,3,4,5]


100 not in lst
```

```python
# Identity
# is and is not


a = [1,2,3]
b = [1,2,3]

print(id(a))
print(id(b))

print(a is b)
```

```python
# Mutable 

lst = ['a' , 'b' , 'c' , 'a' , 'a' , 'a']

lst[2] = 'g'

lst
```

```python
lst = ['Salahuddin' , 'Pooja' , 'Anviti' , 'Abhilash']

'Pooja' in lst
```

```python
lst = ['Salahuddin' , 'Pooja' , 'Anviti' , 'Abhilash']

'Aditi' in lst
```

# Extend


- Insert multiple elements inside a list
- Can be inserted only at the end

```python
lst = ['a' , 'b' , 'c' ]


lst.append(['h' , 'k'])

lst
```

```python
lst = ['a' , 'b' , 'c' ]


lst.extend(['h' , 'k'])

lst
```

# Slicing

- Normal slicing
     a) Zero slicing
     b) User defined slicing
     c) cannot perform slicing on columns
     
     
- Loc
    a) Can perform slicing on rows and columns
    b) Only userdefined indexing is allowed
    c) Start and stop inclusive
    d) dataframe.loc['Row_start_name' : 'Row_stop_name' : step_for_rows , 'Columns_name' : 'Columns_Stop_name': step]

- iloc
    a) Can perform slicing on rows and columns
    b) Only Zero indexing indexing is allowed
    c) Start is inclusive and stop is exclusive
    d) dataframe.iloc['Row_start_index' : 'Row_stop_index' : step_for_rows , 'Columns_index' : 'Columns_index': step]

- Normal slicing
You cannot perform slicing on columns, Only on rows it is possible

     a) Zero slicing
     b) User defined slicing

# Loc
- You can perform slicing on rows and columns
- Only userdefined indexing and slicing is allowed

[rows , columns]

```python
new_df.loc['row_2' : 'row_5' , 'starttime' : 'bikeid']
```

```python
new_df
```

```python
# starttime , start station id , end station id , bikeid


# row_7 to row_2

new_df.loc['row_7':'row_2':-1 , 'starttime':'bikeid': 2 ]
```

```python
df.loc[5:10 , ::2]
```

```python
# 100 - 200 loc
df.loc[100:200  , 'starttime'::2]
```

# iloc
- Slicing on columns and on rows is possible
- Only Zero indexing is allowed 

- Start --> inclusive
- Stop --> Exclusive

```python
new_df.iloc['row_1' : 'row_5']
```

```python
new_df.iloc[2:7 , 1:8]
```

# Adding Columns and deleting columns

```python
df = pd.read_csv('datasets/international_org.csv')
df
```

```python
df
```

# apply

```python
def funct(x):
    return x/5
```

```python
df['new_col_4'] = df['new_col_3'].apply(lambda x : x/5)
```

```python
df
```

```python
df['new_col_5'] = df['Organization Name'].apply(lambda x: x[0])
```

```python
df['new_col_6'] = df['new_col_3'].apply(lambda x : 'Even' if x%2==0 else 'Odd')
```

```python
df
```

# Assign

```python
df = df.assign(col_7 = lambda x: x.new_col_3 * x.new_col_4)
```

```python
df
```

```python
df['new_col_8']  = list(map((lambda x,y: x if x>y else y) , df['new_col_2'] , df['new_col_3']))
```

```python
df
```

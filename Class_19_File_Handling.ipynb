# File handling

Point to remember while working with file handling

1. We have to open the file in order to read or write

syntax:-
    var = open('path/file_name.ext' , 'mode')
    var.close() 
    
    Note:- Closing the file is important orelse the changes will not reflect
    
2. Data read from the file will always be in a form of string

3. Mode of operations -- Text files

- read mode --> 'r'
- write mode --> 'w'
- append mode ---> 'a'
- read and write mode ---> 'r+'
- write and read mode ---> 'w+'
- append and read mode ----> 'a+'

4. Mode of operations -- Binary files

- read mode --> 'rb'
- write mode --> 'wb'
- append mode ---> 'ab'
- read and write mode ---> 'r+b'
- write and read mode ---> 'w+b'
- append and read mode ----> 'a+b'

5. methods on file object
- read() ---> To read the whole data from the file
- write() ---> To write the content on the file
- seek() ---> Used to move the cursor to any specific index location
- tell() --> To tells the position of the cursor

# Read mode -- 'r'

- If the file does not exist it'll throw an error FileNotFound
- If the file exist it'll open the file and place the cursor at the begining of the file
- You can only read from from the file not write into it 

```python
# If the file does not exist it'll throw an error FileNotFound


f = open('file_handling/read_mode.txt' , 'r')
```

```python
# If the file exist it'll open the file and place the cursor at the begining of the file


f = open('file_handling/read_mode.txt' , 'r')


print(f.tell())
```

```python

f = open('file_handling/read_mode.txt' , 'r')


print(f.tell())

a = f.read()


print(a , type(a))
```

```python

f = open('file_handling/read_mode.txt' , 'r')


print(f.tell())

f.write('New_data')
```

# Write mode --> 'w'

- If the file does not exist it'll create a file with that name and extension
- If the file exist it'll open the file and delete the previous data
- You can only write into the file not read from it 

```python
# If the file does not exist it'll create a file with that name and extension


f = open('file_handling/write_mode.sql' , 'w')
```

```python
f = open('file_handling/write_mode.txt' , 'w')


f.write('Hello world')

f.close()
```

```python
f = open('file_handling/write_mode.txt' , 'w')


f.write('Hello world')

f.seek(6)

f.write('New_insert')

f.close()
```

# Methods on read function

- read(char) --> Reads the specified no of char from the current location
- readline() --> read the current line from the current position
- readlines() --> read the whole data and return a list where each element is one line of you data

### Filling the null values

```python
df.fillna(100000)
```

```python
df.fillna({'temperature' : 100 , 
          'windspeed' : 200 , 
          'event' : 'No event'})
```

```python
# 

import seaborn as sns

df = sns.load_dataset('titanic')

df
```

# Task

- % of males survived ? people survived / no of male * 100
- % female many female survived ? 

```python
total_no_males = len(df.loc[df['sex'] == 'male'])
total_no_males
```

```python
total_no_females = len(df.loc[df['sex'] == 'female'])
total_no_females
```

```python
g = df.groupby('sex')

len(g.get_group('female').groupby('survived').get_group(1)) / total_no_females
```

```python
g = df.groupby('sex')

len(g.get_group('male').groupby('survived').get_group(1)) / total_no_males
```

```python
sns.barplot(x = 'sex' , y = 'survived' , data = df)
```

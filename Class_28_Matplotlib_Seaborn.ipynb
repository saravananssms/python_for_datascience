# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
week_days = [1,2,3,4,5,6,7]
temp = [32,34,35,36,32,30,35]
```

# Line graph

```python
plt.plot(week_days , temp)
```

```python
plt.plot(week_days , temp)

plt.xlabel('DAYS')
plt.ylabel('TEMP')
plt.title('Weather information')
```

```python
plt.plot(week_days , temp , c = 'g' , marker = 'o' , ls = '--')

plt.xlabel('DAYS')
plt.ylabel('TEMP')
plt.title('Weather information')
```

# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

fmt = '[marker][line][color]'

```python
plt.plot(week_days , temp , '-.r>' , markersize = 10.5)

plt.xlabel('DAYS')
plt.ylabel('TEMP')
plt.title('Weather information')
```

```python
days=[1,2,3,4,5,6,7] # weekdays
max_t=[50,51,52,48,47,49,46] # max temp recorder at that day
min_t=[43,42,40,44,45,35,37] # min temp
avg_t=[45,48,48,46,40,42,41] # avg temp
```

```python
plt.plot(days , max_t)
plt.plot(days , min_t)
plt.plot(days , avg_t)
```

```python
plt.plot(days , max_t , label = 'MAX')
plt.plot(days , min_t , label = 'MIN')
plt.plot(days , avg_t , label = 'AVG')

plt.legend()
```

```python
plt.plot(days , max_t , label = 'MAX')
plt.plot(days , min_t , label = 'MIN')
plt.plot(days , avg_t , label = 'AVG')

plt.legend(loc = 'lower left')
```

```python
plt.plot(days , max_t , label = 'MAX')
plt.plot(days , min_t , label = 'MIN')
plt.plot(days , avg_t , label = 'AVG')

plt.legend(loc = 1)
plt.grid()
```

# Histogram

```python
blood_sugar = [113,85,90,150,149,88,93,115,135,80,77,82,129]


# 80 - 100 ---> normal 
# 100 - 125 ---> prediabetic
# 125 and above ---> diabetic
```

```python
plt.hist(blood_sugar) # bin = None
```

```python
plt.hist(blood_sugar , bins = [77 , 100 , 125 , 150] , color = 'g' , histtype  = 'step') # bin = None

plt.ylabel('COUNT')
```

# Pie

```python
monthly_expense = [15500 , 6000 , 3000 , 1000 , 500 , 600]
l = ['Rent' , 'Food' , 'Fuel' , 'Internet' , 'phone' , 'others']
```

```python
plt.pie(monthly_expense)

plt.show()
```

```python
plt.pie(monthly_expense , labels= l)

plt.show()
```

```python
plt.pie(monthly_expense , labels= l , radius= 2)

plt.show()
```

```python
plt.pie(monthly_expense , labels= l , radius= 2 , autopct = "%0.2f%%")

plt.show()
```

```python
plt.pie(monthly_expense , labels= l , radius= 1.5 , autopct = "%0.3f%%" , explode = [0,0.45,0,0,0.50,0])

plt.show()
```

```python
plt.pie(monthly_expense , labels= l , radius= 1.5 , autopct = "%0.3f%%"  , explode = [0,0,0.5,0,0,0])

plt.show()
```

```python
monthly_expense = [15500 , 6000 , 3000 , 1000 , 500 , 600]
l = ['Rent' , 'Food' , 'Fuel' , 'Internet' , 'phone' , 'others']
```

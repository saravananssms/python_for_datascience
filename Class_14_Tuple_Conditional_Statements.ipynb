# Properties of tuple 



- It is a seq data tuple
- +ve and -ve indexing
- You can perform indexing and slicing
- Basic op - concatination , repetation , membership , identity
- Allows duplication



### Note:- It is an immutable data type

```python
a = (34,45,56,67)


# +ve
print(a[1])


# -ve
print(a[-2])
```

```python
a = (34,45,56,67,34,45,56)


# +ve - 45,56,67
print(a[1:4])


# -ve - 67 , 56, 45
print(a[3:0:-1])
```

```python
# Concatination
# op --> +
# operands --> Both should be a tuple

a = (1,2,3)
b = (4,5,6)


print(a+b)
```

```python
# Repetation
# op --> *
# operands --> one should be an int and other should be a tuple

a = (1,2,3)
b = 3


print(a * b)
```

```python
# Write a Python function to find the person with the highest salary from a list of nested dictionaries.

employees = [
    {"name": "Alice", "details": {"age": 30, "salary": 50000}},
    {"name": "Bob", "details": {"age": 35, "salary": 60000}},
    {"name": "Charlie", "details": {"age": 28, "salary": 55000}},
]        

s = 0

for i in employees:
    if i['details']['salary'] > s:
        s = i['details']['salary']
for i in employees:
    if s == (i['details']['salary']):
        print(i['name'])
```
